```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data={"product":["laptop","phone","tablet","headphone","monitor","keyboard","mouse","laptop","phone","monitor"],
      "category":["electronics","electronics","electronics","accessories","electronics","accessories","accessories","electronics","electronics","electronics"],
      "sales":[75000,50000,30000,5000,25000,3000,1500,80000,55000,28000],
      "quantity":[3,5,4,10,6,15,20,4,6,7]}
df=pd.DataFrame(data)
print(df)
    
```

# Sales Data Analysis Dashboard
Prepared using Python,Pandas and Matplotlib

```python
print("========== BUSINESS SUMMARY ==========")
print("total sales:",df["sales"].sum())
print("high sales:",df["sales"].max())
print("low sales:",df["sales"].min())
```

```python
print("average sales:",df["sales"].mean())
print("count of products:",df["product"].count())
print("total quantity:",df["quantity"].sum())
```

```python
df
```

```python
plt.figure(figsize=(12,3))
plt.bar(df["product"],df["sales"])
plt.title("sales by product")
plt.xlabel("product")
plt.ylabel("sales")
plt.show()
```

```python
plt.figure(figsize=(12,5))
plt.plot(df["product"],df["sales"],marker='o')
plt.title("sales by trend")
plt.xlabel("products")
plt.ylabel("sales")
plt.show()
```

```python
category_sales=df.groupby("category")["sales"].sum()
plt.figure(figsize=(6,6))
plt.pie(category_sales,labels=category_sales.index,autopct="%1.1f%%")
plt.title("category wise sales")
plt.show()
```

```python
plt.figure(figsize=(12,4))
plt.hist(df["sales"],bins=5)
plt.title("sales distribution")
plt.xlabel("sales")
plt.ylabel("distribution")
plt.show()
```
