```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "product":["Laptop","Phone","Tablet","Headphone","Monitor","Keyboard","Mouse","Laptop","Phone","Monitor"],
    "category":["Electronics","Electronics","Electronics","Accessories","Electronics","Accessories","Accessories","Electronics","Electronics","Electronics"],
    "sales":[75000,50000,30000,5000,25000,3000,1500,80000,55000,20000],
    "quantity":[3,5,4,10,6,15,20,4,6,7]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
df.info()
```

```python
df.describe()
```

```python
plt.figure(figsize=(6,4))
plt.bar(df["product"], df["sales"])
plt.title("Sales by Product")
plt.xlabel("Products")
plt.ylabel("Sales (₹)")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.plot(df["product"], df["sales"], marker="o")
plt.title("Sales Trend")
plt.xlabel("Products")
plt.ylabel("Sales (₹)")
plt.show()plt.figure(figsize=(6,6))
plt.pie(df["sales"], labels=df["product"], autopct="%1.1f%%")
plt.title("Sales Distribution")
plt.show()

```

```python
plt.figure(figsize=(6,6))
plt.pie(df["sales"], labels=df["product"], autopct="%1.1f%%")
plt.title("Sales Distribution")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.hist(df["sales"], bins=5)
plt.title("Sales Distribution Histogram")
plt.xlabel("Sales (₹)")
plt.ylabel("Frequency")
plt.show()
```
