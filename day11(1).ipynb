```python
import pandas as pd
```

```python
data = {
    "Product": ["Laptop","Phone","Tablet","Headphones","Monitor",
                "Keyboard","Mouse","Laptop","Phone","Monitor"],
    "Category": ["Electronics","Electronics","Electronics","Accessories",
                 "Electronics","Accessories","Accessories","Electronics",
                 "Electronics","Electronics"],
    "Sales": [75000,50000,30000,5000,25000,3000,1500,80000,55000,28000],
    "Quantity": [3,5,4,10,6,15,20,4,6,7]
}

df = pd.DataFrame(data)
```

```python
print("Total Sales :", df["Sales"].sum())
print("Average Sales :", df["Sales"].mean())
print("Highest Sales :", df["Sales"].max())
print("Lowest Sales :", df["Sales"].min())
```

```python
print(df.loc[df["Sales"].idxmax()])
```

```python
print(df.loc[df["Sales"].idxmin()])
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

print(category_sales)
```

```python
print("Total Quantity Sold :", df["Quantity"].sum())
```
